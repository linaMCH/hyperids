```python
#cherger l'entrainement officiel 
import gc

base = '/kaggle/input/datasets/mrwellsdavid/unsw-nb15/'

del df, a, b
gc.collect()

train = pd.read_csv(base + 'UNSW_NB15_testing-set.csv')  # partition d'entraînement officielle
print(train.shape)
print('valeurs manquantes :', train.isna().sum().sum())
print('doublons (hors id) :', train.drop(columns=['id']).duplicated().sum())
```

```python
#variables de type texte et cible 
for c in ['proto', 'service', 'state']:
    print(c, ':', train[c].nunique(), 'valeurs différentes')

print()
print(train['service'].value_counts().head(10))
print()
print(train['label'].value_counts(normalize=True))
```

```python
#resume statistique 
display(train.describe().T)
```

```python
#veirfier que attack_cat contient la reponse 
display(pd.crosstab(train['attack_cat'], train['label']))
```

```python
#mesurer les doublons 
cols = [c for c in train.columns if c not in ['id', 'attack_cat', 'label']]

# doublons sur les caractéristiques seules
dup_mask = train.duplicated(subset=cols, keep='first')
print('lignes en doublon (caractéristiques) :', dup_mask.sum())

# répartition des doublons par classe
print()
print('doublons par label :')
print(train.loc[dup_mask, 'label'].value_counts())
print()
print('doublons par attack_cat :')
print(train.loc[dup_mask, 'attack_cat'].value_counts())

# caractéristiques identiques mais étiquettes contradictoires ?
g = train.groupby(cols)['label'].nunique()
print()
print('groupes avec étiquettes contradictoires :', (g > 1).sum())

# répartition après suppression des doublons
dedup = train.drop_duplicates(subset=cols, keep='first')
print()
print('lignes après suppression :', dedup.shape[0])
print(dedup['label'].value_counts(normalize=True))
print(dedup['attack_cat'].value_counts())
```

```python
#les deux colonnes suspectes et les deux categories 
print('ct_ftp_cmd identique à is_ftp_login :', (train['ct_ftp_cmd'] == train['is_ftp_login']).mean())
print()
print(train['is_ftp_login'].value_counts())
print()
print(train['proto'].value_counts().head(10))
print()
print(train['state'].value_counts())
```

```python
#afin de decider du regroupement 
# 1. lignes concernées par les étiquettes contradictoires
n_labels = train.groupby(cols)['label'].transform('nunique')
conflict = train[n_labels > 1]
print('lignes dans des groupes contradictoires :', len(conflict))
print(conflict['attack_cat'].value_counts())

# 2. taux d'attaque par protocole (entraînement)
top = train['proto'].value_counts().head(10).index
proto_group = train['proto'].where(train['proto'].isin(top), 'autres')
tab = train.groupby(proto_group)['label'].agg(['count', 'mean'])
tab.columns = ['lignes', 'taux_attaque']
print(tab.sort_values('lignes', ascending=False))
```
